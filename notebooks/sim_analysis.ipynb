```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import matplotlib as mpl
from matplotlib.ticker import MultipleLocator
from cmcrameri import cm

import sys
sys.path.append("../src")

from crop_model_calibration_and_run import summarize_productive_grids
```

```python
plt.rcParams.update({
    "font.family": "sans-serif",
    "font.sans-serif": ["Arial", "Helvetica"],
    "font.size": 11,
    "axes.linewidth": 1,
    "axes.labelsize": 12,
    "axes.titlesize": 13,
    "xtick.direction": "out",
    "ytick.direction": "out",
    "xtick.major.width": 1,
    "ytick.major.width": 1,
})
```

```python
grid_yield_path = Path(
    "../../Results/3.2_SimulatedYield/"
    "aquacrop_rice_grid_yield_ph_ythr1pct_0901_0p3_var_260714.csv"
)

country_ensemble_yield_path = Path(
    "../../Results/3.2_SimulatedYield/"
    "aquacrop_rice_country_ensemble_yield_"
    "ph_ythr1pct_0901_0p3_var_260714.csv"
)
output_plot_path = (
    Path("../../fig") / "yield_ph_ythr1pct_0901_0p3_var.svg"
)

yield_threshold = 0.1
simulation_results_df = pd.read_csv(grid_yield_path)
comparison_df = pd.read_csv(country_ensemble_yield_path)
```

```python
summary_all_df = summarize_productive_grids(
    simulation_results_df,
    yield_threshold=0
)

summary_threshold_df = summarize_productive_grids(
    simulation_results_df,
    yield_threshold=0.1
)

summary_ratio_df = (
    summary_all_df.merge(
        summary_threshold_df,
        on="year",
        suffixes=("_all", "_threshold"),
    )
    .assign(
        grid_ratio=lambda x: x["n_gid_threshold"] / x["n_gid_all"],
        harvest_area_ratio=lambda x: (
            x["harvest_area_ha_threshold"]
            / x["harvest_area_ha_all"]
        ),
    )
    .loc[:, ["year", "grid_ratio", "harvest_area_ratio"]]
)

incomplete_ratio_df = summary_ratio_df[
    (summary_ratio_df["grid_ratio"] < 1)
    | (summary_ratio_df["harvest_area_ratio"] < 1)
]

print(f"No. of years with complete-loss grids: {len(incomplete_ratio_df)}")
print(
    f"Range of grid coverage: "
    f"{incomplete_ratio_df["grid_ratio"].min():.1%}~"
    f"{incomplete_ratio_df["grid_ratio"].max():.1%}"
)
print(
    f"Range of HA coverage: "
    f"{incomplete_ratio_df["harvest_area_ratio"].min():.1%}~"
    f"{incomplete_ratio_df["harvest_area_ratio"].max():.1%}"
)
incomplete_ratio_df
```

```python
years = comparison_df.year
y = comparison_df.yield_obs_t_ha
y_simulated = comparison_df.yield_sim_t_ha

msr = mean_squared_error(y, y_simulated)
r2 = r2_score(y, y_simulated)
print(f"RMSE: {np.sqrt(msr):.3f}")
print(f"R-squared score: {r2:.2f}")

print(f"\nMean y: {np.mean(y):.2f} t/ha")
print(f"Range of y: {y.min():.2f}~{y.max():.2f} t/ha")
print(f"Mean y_simulated: {np.mean(y_simulated):.2f} t/ha")
print(
    f"Range of y_simulated: "
    f"{y_simulated.min():.2f}~{y_simulated.max():.2f} t/ha"
)
```

```python
series_1 = summary_ratio_df.grid_ratio*100
series_2 = summary_ratio_df.harvest_area_ratio*100

cmap = cm.grayC

norm = mpl.colors.Normalize(vmin=0, vmax=100)

ymin = np.floor(min(y.min(), y_simulated.min()) * 10) / 10
ymax = np.ceil(max(y.max(), y_simulated.max()) * 10) / 10

# -------------------------
# Figure with subplots
# -------------------------
fig, (ax, ax1, ax2) = plt.subplots(
    3, 1,
    figsize=(4, 1.8),
    sharex=True,
    gridspec_kw={
        "height_ratios": [3, 0.3, 0.3],
        "hspace": 0.05,   # controls vertical spacing
    },
)

# -------------------------
# Main plot (yield)
# -------------------------
for gid in simulation_results_df.GID.unique():
    gid_simulation_df = simulation_results_df[
        simulation_results_df["GID"]==gid
        ]
    ax.plot(
        gid_simulation_df.year,
        gid_simulation_df.yield_sim_t_ha,
        color="#0072B2",
        lw=0.3,
        alpha=0.2,
        zorder=1
    )

ax.axhline(
    y=np.mean(y),
    color="#000000",
    linestyle="--",
    lw=1,
    alpha=0.8
)
ax.axhline(
    y=np.mean(y_simulated),
    color="#009E73",
    linestyle="--",
    lw=1,
    alpha=0.8
)

ax.plot(
    years,
    y,
    color="#000000",
    lw=1.6,
    label="FAO-detrended",
    zorder=2
)
ax.plot(
    years,
    y_simulated,
    color="#009E73",
    lw=1.6,
    label="HA-weighted ensemble",
    zorder=3
)

ax.set_ylim(1.8,2.9)
ax.set_ylabel("Yield (t/ha)")
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# -------------------------
# Color bar 1
# -------------------------
ax1.imshow(
    [series_1],
    aspect="auto",
    cmap=cmap,
    norm=norm,
    extent=[years.min()-0.5, years.max()+0.5, 0, 1],
)

ax1.set_yticks([])
ax1.set_ylabel("% grids", rotation=0, labelpad=20, va="center")

for spine in ax1.spines.values():
    spine.set_visible(False)

# -------------------------
# Color bar 2
# -------------------------
grid_image = ax2.imshow(
    [series_2],
    aspect="auto",
    cmap=cmap,
    norm=norm,
    extent=[years.min()-0.5, years.max()+0.5, 0, 1],
)

ax2.set_yticks([])
ax2.set_ylabel("% HA", rotation=0, labelpad=20, va="center")

for spine in ax2.spines.values():
    spine.set_visible(False)

# -------------------------
# Shared x-axis formatting
# -------------------------
ax2.set_xlim(1991-0.5, 2020+0.5)

#plt.show()
plt.savefig(output_plot_path, bbox_inches="tight", dpi=300)
```
