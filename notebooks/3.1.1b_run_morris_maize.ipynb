```python
from pathlib import Path
import pickle
import numpy as np

import sys
sys.path.append("../src")

from morris_sensitivity import run_morris_analysis
```

## Case MM-maize

```python
# Default crop parameters for rice
crop_params = {
    "CCx": 0.96,
    "CGC": 0.012494,
    "CDC": 0.01,
    "Zmax": 2.3,
    "WP": 33.7,
    "WPy": 100.0,
    "HI0": 0.48,
    "p_up1": 0.14,
    "p_up2": 0.69,
    "p_up3": 0.69,
    "fshape_w1": 2.9,
    "fshape_w2": 6.0,
    "fshape_w3": 2.7,
}

results = run_morris_analysis(
    gid_list=[160391],
    weather_folder="../data/aquacrop_daily_weather_inputs_processed/mm/",
    crop_params=crop_params,
    crop_name="MaizeGDD",
    planting_date="05/01",
    sim_start_date="1991/01/01",
    sim_end_date="2020/12/31",
    soil_type="ClayLoam",
    n_outputs=30,
    lower_factor=0.4,
    upper_factor=1
)

Si_list = results["indices"]
model_outputs = results["outputs"]
parameter_samples = results["samples"]
problem_definition = results["problem"]
```

```python
output_file_path = Path(
    "../outputs/3.1_morris_sensitivity_summaries/"
    "morris_sensitivity_summary_maize_mm_left_0p6_var_20260716.pkl"
)

output_file_path.parent.mkdir(parents=True, exist_ok=True)

with output_file_path.open("wb") as file_handle:
    pickle.dump(results, file_handle)
```

```python
# check effect: positive or negative
stacked = np.ma.array([Si['mu'] for Si in Si_list])
stacked.mean(axis=0)
```
